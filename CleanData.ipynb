```python
import numpy as np
import pandas as pd

pd.set_option('display.max_columns', None)
print('Libraries imported successfully!')
```

```python
raw_filename = 'default of credit card clients.csv'
df = pd.read_csv(raw_filename)

print(f'Raw dataset shape: {df.shape[0]} rows, {df.shape[1]} columns')
df.head()
```

```python
missing_values = df.isnull().sum()
total_missing = missing_values.sum()
print(f'Total missing values across the dataset: {total_missing}')
```

```python
feature_cols = [col for col in df.columns if col != 'ID']
duplicate_count = df.duplicated(subset=feature_cols).sum()
print(f'Duplicate entries found: {duplicate_count}')

# Drop duplicates
df_cleaned = df.drop_duplicates(subset=feature_cols).copy()
print(f'Shape after removing duplicates: {df_cleaned.shape[0]} rows')
```

```python
df_cleaned['EDUCATION'] = df_cleaned['EDUCATION'].replace({0: 4, 5: 4, 6: 4})

df_cleaned['MARRIAGE'] = df_cleaned['MARRIAGE'].replace({0: 3})

print('EDUCATION distribution:', df_cleaned['EDUCATION'].value_counts().to_dict())
print('MARRIAGE distribution:', df_cleaned['MARRIAGE'].value_counts().to_dict())
```

```python
if 'ID' in df_cleaned.columns:
  df_cleaned = df_cleaned.drop(columns=['ID'])
  print("'ID' column dropped.")

print(
    f'Final feature matrix shape: {df_cleaned.shape[0]} rows, {df_cleaned.shape[1]} columns'
)
```

```python
output_filename = 'default of credit card clients (cleaned).csv'
df_cleaned.to_csv(output_filename, index=False)
print(f"Exported to '{output_filename}' successfully!")
```
