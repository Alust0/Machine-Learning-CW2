# Part D Development of 3 base models

## classification: Gradient boosting

```python
import pandas as pd

#loading clean data 
train_df = pd.read_csv('credit_train_preprocessed.csv')
test_df = pd.read_csv('credit_test_preprocessed.csv')


X_train_resampled = train_df.drop(columns=['dpnm'])
y_train_resampled = train_df['dpnm']

X_test = test_df.drop(columns=['dpnm'])
y_test = test_df['dpnm']

print("Data loaded successfully!")
print("X_train_resampled shape:", X_train_resampled.shape)
print("X_test shape:", X_test.shape)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report, roc_auc_score, f1_score

# gradient boosting with default parameters
gb_baseline = GradientBoostingClassifier(random_state=42)

# training on resampled data
gb_baseline.fit(X_train_resampled, y_train_resampled)

# predicting and evaluating
y_pred_base = gb_baseline.predict(X_test)
y_prob_base = gb_baseline.predict_proba(X_test)[:, 1]

print("Gradient Boosting")
print(f"ROC-AUC Score: {roc_auc_score(y_test, y_prob_base):.4f}")
print(f"F1-Score:      {f1_score(y_test, y_pred_base):.4f}\n")
print(classification_report(y_test, y_pred_base))
```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score, f1_score

# 1. Initialize Random Forest with default parameters
rf_baseline = RandomForestClassifier(random_state=42)

# training the sample data
rf_baseline.fit(X_train_resampled, y_train_resampled)

# predict and eveliate on untouched test
y_pred_rf = rf_baseline.predict(X_test)
y_prob_rf = rf_baseline.predict_proba(X_test)[:, 1]

print("Random forest results")
print(f"ROC-AUC Score: {roc_auc_score(y_test, y_prob_rf):.4f}")
print(f"F1-Score:      {f1_score(y_test, y_pred_rf):.4f}\n")
print(classification_report(y_test, y_pred_rf))
```
