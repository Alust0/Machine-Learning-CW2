# Part E: Hyperparameter optimisation

```python
import pandas as pd

train_df = pd.read_csv('credit_train_preprocessed.csv')
test_df = pd.read_csv('credit_test_preprocessed.csv')

# Separate features (X) and target (y)
X_train_resampled = train_df.drop(columns=['dpnm'])
y_train_resampled = train_df['dpnm']

X_test = test_df.drop(columns=['dpnm'])
y_test = test_df['dpnm']

print("Data is loaded")
```

```python
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
import pandas as pd
import joblib


# 5 fold stratefied CV
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scoring_metric = 'roc_auc'

#Random Forest
rf_params = {
    'n_estimators': [100, 200, 300],
    'max_depth': [6, 10, 15, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}
rf_search = RandomizedSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_distributions=rf_params,
    n_iter=10,
    cv=cv_strategy,
    scoring=scoring_metric,
    random_state=42,
    n_jobs=-1
)
rf_search.fit(X_train_resampled, y_train_resampled)
print("Random forest best params")
print(rf_search.best_params_)
print(f"Best CV ROC-AUC: {rf_search.best_score_:.4f}\n")


# Gradient boosting
gb_params = {
    'n_estimators': [100, 150, 200],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 4, 5],
    'subsample': [0.8, 1.0]
}
gb_search = RandomizedSearchCV(
    estimator=GradientBoostingClassifier(random_state=42),
    param_distributions=gb_params,
    n_iter=10,
    cv=cv_strategy,
    scoring=scoring_metric,
    random_state=42,
    n_jobs=-1
)

gb_search.fit(X_train_resampled, y_train_resampled)
print("Gradient boosting")
print(gb_search.best_params_)
print(f"Best CV ROC-AUC: {gb_search.best_score_:.4f}\n")

#saving for part f 
joblib.dump(rf_search.best_estimator_, 'best_rf.joblib')
joblib.dump(gb_search.best_estimator_, 'best_gb.joblib')
print("Saved best_rf.joblib and best_gb.joblib")
```
